# Partie 3 – Découpage Train/Test

Diviser X en deux ensembles distincts : un pour l'entraînement (train) et un pour le test (test). 
Avec les conditions suivantes : 20% des données serviront au test ; garantir la reproductibilité du 
découpage ; conserver les mêmes proportions de classes dans l'ensemble de train et de test que 
dans les données d'origine. 

```python
df = df.dropna(subset=["etat"])
```

```python
print(df["etat"].isnull().sum())
```

```python
x = df[
    [
        "temperature",
        "humidite",
        "pression",
        "consommation"
    ]
]
```

```python
# Definissons y la variable a predire
y = df["etat"]
```

```python
print(x.shape)
print(y.shape)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
```

```python
y.value_counts(normalize=True)
```

```python
y_train.value_counts(normalize=True)
```

# Partie 4 – Gestion des valeurs manquantes

1) Vérifier l’existence de valeurs manquantes 

```python
x.isnull().sum()
```

2) Sélectionner SimpleImputer avec la médiane 

```python
imputer = SimpleImputer(
    strategy="median"
)
```

3) Qu’est ce qui justifie le choix de la médiane ?

La médiane est particulièrement utile lorsque les données contiennent des :
- valeurs extrêmes ;
- anomalies ;
- outliers.

Elle est moins influencée par les valeurs extrêmes que la moyenne.

4) Trouver les paramètres (médianes) de l’imputeur sur X_train

```python
imputer.fit(x_train)
```

```python
# Affichage des medianes calcules 
mediane_imputer = pd.Series(imputer.statistics_,index=x_train.columns)
print(mediane_imputer)
```

5) Déterminer X_train_imputed et X_test_imputed, les transformés de X_train et X_test 

```python
x_train_imputed = imputer.transform(x_train)  #Le transformé de x_train
```

```python
x_test_imputed = imputer.transform(x_test)  # Le transformé de x_test
```

# Partie 5 – Mise à l'échelle 

1) Sélectionner StandardScaler pour mettre à l’échelle les transformés de l’imputation

```python
scaler = StandardScaler()
```

```python
scaler.fit(x_train_imputed) # Entrainement du scaler
```

2) Qu’est ce qui justifie la standardisation ? 

Sans standardisation, la variable ayant les valeurs numériques les plus grandes peut influencer davantage le calcul de distance.
La standardisation met donc les variables sur une échelle comparable.

3) Trouver les paramètres (moyennes et écart-types) du scaleur sur X_train_imputed 

```python
parametres_scaler = pd.DataFrame({
    "Variable": x_train.columns,
    "Moyenne": scaler.mean_,
    "Ecart_type": scaler.scale_
})

parametres_scaler
```

4) Déterminer X_train_scaled et X_test_scaled, les transformés de X_train_imputed et 
X_test_imputed 

```python
x_train_scaled = scaler.transform(x_train_imputed)
```

```python
x_test_scaled = scaler.transform(x_test_imputed)
```

# Partie 6 – Entrainement et prédiction d’un modèle

1) Sélectionner le modèle KNN (k plus proches voisins) avec 5, le nombre de voisins à prendre en 
compte 

```python
knn = KNeighborsClassifier(n_neighbors=5)  # Creation du modele
```

2) Entrainer le modèle

```python
knn.fit(x_train_scaled,y_train) #Entrainement du modele
```

3) Déterminer y_pred, la prédiction du modèle avec l’ensemble de test

```python
y_pred = knn.predict(x_test_scaled)  # Prediction du modele
```

4) Afficher quelques prédictions 


```python
print(y_pred[:10]) # Affichage de quelques predictions
```

5) Comparer ces prédictions avec les vraies valeurs 

```python
comparaison = pd.DataFrame({"Valeur_reelle": y_test.values[:10],"Prediction": y_pred[:10]})
comparaison    # Comparaison avec les vraies valeurs
```

# Partie 7 – Evaluation du modèle 

1) Evaluer le modèle avec Accuracy puis afficher le résultat 

```python
accuracy = accuracy_score(y_test,y_pred)
print("Accuracy :",accuracy)
```

```python
print(f"Accuracy : {accuracy:.2%}")
```

2) Pourquoi Accuracy ne suffit pas toujours pour évaluer ?

Si les classes sont déséquilibrées, (OK :900 observations,ALERTE:80 observations, ERREUR:20 observations) un modèle peut avoir une bonne accuracy tout en prédisant mal une classe minoritaire.

Un modèle qui prédit presque toujours OK pourrait avoir une accuracy élevée, mais serait mauvais pour détecter les ERREUR.

3) Déterminer, afficher et commenter la matrice de confusion 

```python
cm = confusion_matrix(y_test,y_pred)
print(cm)

```

```python
classes = knn.classes_
cm_df = pd.DataFrame(cm,index=classes,columns=classes)
cm_df         #Les lignes représentent les vraies classes et les colonnes les classes prédites.
```

Commentaires:
- Pour la  Classe OK
Il y avait 113 mesures réellement en état OK.
Le modèle a prédit :
113 comme OK 
0 comme ALERTE
0 comme ERREUR
Conclusion: Le modèle reconnaît donc parfaitement les observations de la classe OK dans cet ensemble de test.

- Pour la Classe ALERTE
Il y avait 6 mesures réellement en ALERTE :
2 + 0 + 4 = 6
Parmi elles :
2 ont été correctement prédites comme ALERTE 
4 ont été incorrectement prédites comme OK 
0 ont été prédites comme ERREUR
Conclusion: Le modèle a donc des difficultés à détecter les capteurs en état ALERTE.

- Pour la Classe ERREUR
Il y avait 1 mesure réellement en ERREUR.
Le modèle a prédit cette mesure comme :
ALERTE : 1
ERREUR : 0
OK : 0
Conclusion:  Le modèle n'a correctement identifié aucune observation de la classe ERREUR.

Cela montre une difficulté importante à reconnaître cette classe.

4) Visualiser cette matrice avec Seaborn 

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=classes,
    yticklabels=classes
)

plt.title("Matrice de confusion")
plt.xlabel("Classe prédite")
plt.ylabel("Classe réelle")

plt.show()
```

Commentaires: Le concept de lecture de cette matrice repose sur ce principe selon lequel :plus les valeurs sont concentrées sur la diagonale, meilleur est le modèle.

5) Afficher et commenter le rapport de classificaton 

```python
print(classification_report(y_test,y_pred))
```
